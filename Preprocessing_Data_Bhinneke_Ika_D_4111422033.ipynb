1. Memuat Data

```python
import pandas as pd
```

```python
#membaca file CSV (Comma Separated Values) dan mengubahnya menjadi DataFrame Pandas
df = pd.read_csv('/content/sample_data/movie_sample_dataset.csv')
```

2. Memeriksa Data

```python
# menampilkan beberapa baris pertama dari dataset untuk mengecek struktur data
# default = 5 baris
df.head()
```

```python
# mengetahui ukuran dataset
# hasil (99, 13) artinya jumlah baris = 99 dan jumlah kolom = 13
df.shape
```

3. Membersihkan Data

```python
import numpy as np

# replace "?" menjadi NaN
df.replace("?", np.nan, inplace = True)

# cek banyaknya missing value pada masing-masing kolom
print(df.isnull().sum())

# color = 11 missing value
# director name = 11 missing value
# genres = 1 missing value
# budget = 4 missing value
```

```python
# menghapus baris yang memiliki nilai NaN di kolom gross

# simply drop whole row with NaN in "gross" column
df.dropna(subset=["gross"], axis=0, inplace=True)
# reset index, because we droped two rows
df.reset_index(drop=True, inplace=True)
print(df.head())
```

```python
# menghapus baris yang memiliki nilai NaN di kolom budget

# simply drop whole row with NaN in "budget" column
df.dropna(subset=["budget"], axis=0, inplace=True)
# reset index, because we droped two rows
df.reset_index(drop=True, inplace=True)
print(df.head())
```

```python
# cek ulang banyaknya missing value pada masing-masing kolom
print(df.isnull().sum())
```

```python
# jumlah cek baris dan kolom setelah missing value dihilangkan
df.shape
```

```python
# mengatasi nilai yg tdk konsisten atau kesalahan penulisan di kolom.
# mengganti kata 'Color' mjd 'color'.

df['color'] = df['color'].str.strip().str.lower()
print(df["color"].value_counts())
```

```python
#menghapus nilai-nilai tidak standar, seperti N/A
df = df.dropna()
print(df.isnull().sum())
```

4. Transformasi Data

```python
#mengubah tipe data kolom menjadi tipe data yg sesuai
df.dtypes
```

```python
#pisahkan genre yg bergabung dlm 1 kolom mjd bbrp kolom terpisah

# buat dummy variable untuk semua genre
df_genres_ohe = df['genres'].str.get_dummies(sep='|')

# gabungkan ke dataframe asli
df_ohe = pd.concat([df, df_genres_ohe], axis=1)

print(df_ohe.head(10))
```

```python
# cek banyaknya missing value pada masing-masing kolom
print(df.isnull().sum())
```

```python
#normalisasi teks untuk memastikan konsistensi
for col in df_ohe.select_dtypes(include='object').columns:
    df_ohe[col] = df_ohe[col].str.strip().str.lower()
    df_ohe.columns = df_ohe.columns.str.strip().str.lower()

print(df_ohe.head(10))
```

```python
#simpan data yg telah dilakukan preprocessing ke dalam file CSV baru
df_ohe.to_csv('movie_dataset_cleaned.csv')
```
